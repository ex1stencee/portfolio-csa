## Popcorn Hack 2.1

**Question:** What happens if a user types "twenty" when nextInt() expects a number? How would you make your program more robust?

**Answer:** An error, specifically java.util.InputMismatchException, would be thrown since "twenty" is a string, while nextInt() expects an integer, and thus cannot read the string. In order to make the program more robust, we can use solutions like hasNextInt() in order to check whether or not the input is an integer, use NextLine() in order to read our input like a string, then test and convert this input into an integer, or use a try-catch block to catch any errors that come with bad inputs. 

## Popcorn Hack 2.2

**Question:** If a user enters "San Diego" when next() is called, what gets stored? What about nextLine()?

**Answer:** next() reads up to the next whitespace, which in this case, means it would only read the "San" part, since there is a space between "San" and "Diego". nextLine() stores all of "San Diego" since it reads the whole line, or everything up to the newline character, \n, which is passed when the user presses their enter key.

## Popcorn Hack 3

**Question:** Trace through the complete example. If the user enters “Alice Wonderland”, 20, and 3.85, what is the exact output?

```java
import java.util.Scanner;

public class InputLesson {
    public static void main(String[] args) {
        System.out.println("=== User Information Program ===\n");
        Scanner sc = new Scanner(System.in);

        System.out.print("Enter your full name: ");
        String name = sc.nextLine(); // Enters Alice Wonderland
        
        System.out.print("Enter your age: ");
        int age = sc.nextInt(); // Enters 20
        
        System.out.print("Enter your GPA: ");
        double gpa = sc.nextDouble(); // Enters 3.85
        
        System.out.println("\n--- Summary ---");
        System.out.println("Name: " + name); // Prints: "Name: Alice Wonderland"
        System.out.println("Age: " + age + " (next year: " + (age + 1) + ")"); // Prints: "Age 20 (next year: 21)"
        System.out.println("GPA: " + gpa); // Prints: GPA: 3.85
        
        sc.close();
    }
}
```

```
=== User Information Program ===
Enter your full name: Alice Wonderland
Enter your age: 20
Enter your GPA: 3.85
--- Summary ---
Name: Alice Wonderland
Age: 20 (next year: 21)
GPA: 3.85
```

```java
int x = 4;

```

**Answer:** B. 7

## Homework Hack 1

**Task:** Modify the example program to:
- Prompt for three integers
- Calculate their average (watch for integer division!)
- Display the result with two decimal places

````java
import java.util.Scanner;

public class InputLesson {
    public static void main(String[] args) {
        System.out.println("=== Three-Number Average Program ===");
        Scanner sc = new Scanner(System.in);

        System.out.print("Enter the first number: ");
        int num1 = sc.nextInt();
        System.out.println(num1);

        System.out.print("Enter the second number: ");
        int num2 = sc.nextInt();
        System.out.println(num2);

        System.out.print("Enter the third number: ");
        int num3 = sc.nextInt();
        System.out.println(num3);
        
        System.out.println("=== Results ===");
        double average = (num1 + num2 + num3) / 3.0;
        System.out.printf("The average of %d, %d, and %d is %.2f", num1, num2, num3, average);
        
        sc.close();
    }
}

InputLesson.main(new String[]{});
````

## Homework Hack 2

The assignment operator (`=`) is used to assign a value to a variable, while the comparison operator (`==`) is to see if two values are equal.

The assignment operator (`=`) is used like so:
```java
int x = 5;
```
The comparison operator (`==`) is used like so:
```java
if (value == 16) {
    System.out.println("You guessed the number!");
}
```

A mix-up between these two could either lead to the failure of a conditional, or the failure to define a variable. In certain situations where the conditional is checking for an already-defined boolean value, replacing the comparison operator with the assignment operator could lead to the success of the conditional, as well as the changing of the variable assignment, which could have severe implications.
