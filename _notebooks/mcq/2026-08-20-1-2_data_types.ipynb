## [Popcorn Hack] Practice #2: Pick the Best Type

**Pick the Best Type:**

1. Number of siblings -> int
2. Your first name -> string
3. Are you hungry? -> boolean
4. Your favorite letter -> char
5. Your height in inches -> double
6. Number of days in a year (never changes) -> final int

```java
int siblings = 2;
String firstName = "Jacob";
boolean isHungry = true;
char favoriteLetter = 'a';
double heightInches = 66;
final int DAYS_PER_YEAR = 365;

System.out.println("Siblings: " + siblings);
System.out.println("Name: " + firstName);
System.out.println("Hungry: " + isHungry);
System.out.println("Favorite letter: " + favoriteLetter);
System.out.println("Height: " + heightInches + " inches");
System.out.println("Days per year: " + DAYS_PER_YEAR);
```

## Homework Hack: Simple Grade Calculator

**Make a program that calculates a student's grade**

```java
final String NAME = "Timmy";
final String CLASS = "AP Computer Science A";
int test1 = 77;
int test2 = 85;
int test3 = 86;

double average = (test1 + test2 + test3)/3.0;

char grade;

if (average >= 90) {
    grade = 'A';
} else if (average >= 80) {
    grade = 'B';
} else if (average >= 70) {
    grade = 'C';
} else if (average >= 60) {
    grade = 'D';
} else {
    grade = 'F';
}

System.out.println("Report Card");
System.out.println("Name: " + NAME);
System.out.println("Class: " + CLASS);
System.out.println("Test Scores: " + test1 + ", " + test2 + ", " + test3);
System.out.println("Final Grade: " + grade);
```
