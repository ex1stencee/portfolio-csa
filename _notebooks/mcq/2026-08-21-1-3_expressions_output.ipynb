## Popcorn Hack 2

**Reference Code:**

````java
import java.util.Scanner;

public class Menu {
    public static void main(String[] args) {
        Scanner sc = new Scanner(System.in);

        System.out.println("==== Main Menu ====");
        System.out.println("1. Start Game");
        System.out.println("2. Instructions");
        System.out.println("3. Exit");

        System.out.print("Choose an option: ");
        int choice = sc.nextInt();

        System.out.println("You selected option: " + choice);

        int optionCount = 3;
        System.out.println("There are " + optionCount + " total options.");
    }
}

Menu.main(new String[]{});
````

**Formatting tools used:**
1. System.out.println() &rarr; Prints text to a new line, good for creating the menu display which takes up multiple line and isolating each text in its own section
2. System.out.print() &rarr; Prints text without a new line, good for input options, allows for the printing of the option to appear on the same line, making it look neater
3. \+ &rarr; Used to concatnate the choice to the text that explained where the choice came from

## Homework Hack 2

**Expand `Menu.java` into a calculator menu**


````java
import java.util.Scanner;

public class Menu {
    public static void main(String[] args) {
        Scanner sc = new Scanner(System.in);

        System.out.println("==== Main Menu ====");
        System.out.println("1. Add");
        System.out.println("2. Subtract");
        System.out.println("3. Multiply");
        System.out.println("4. Divide");

        System.out.println("Choose an option");
        int choice = sc.nextInt();

        System.out.println("You selected option: " + choice);

        System.out.println("==== Enter Numbers ====");
        System.out.print("Enter the first number: ");
        int num1 = sc.nextInt();
        System.out.print(num1 + "\n");
        System.out.print("Enter the second number: ");
        int num2 = sc.nextInt();
        System.out.print(num2 + "\n");

        if (choice == 1) {
            System.out.println(num1 + " + " + num2 + " = " + (num1 + num2));
        } else if (choice == 2) {
            System.out.println(num1 + " - " + num2 + " = " + (num1 - num2));
        } else if (choice == 3) {
            System.out.println(num1 + " * " + num2 + " = " + (num1 * num2));
        } else if (choice == 4) {
            System.out.println(num1 + " / " + num2 + " = " + (num1 / num2));
        } else {
            System.out.println("Please enter a valid choice");
        }
    }
}

Menu.main(new String[]{});
````
