## Popcorn Hack 1

```java
import java.util.ArrayList;

public class CodeRunner7_PopcornHack1 {
    public static void main(String[] args) {

        double power = Math.pow(3, 4);
        System.out.println("3^4 = " + power);

        double squareRoot = Math.sqrt(64);
        System.out.println("Square root of 64 = " + squareRoot);

        ArrayList<String> colors = new ArrayList<String>();

        colors.add("red");
        colors.add("blue");
        colors.add("green");

        System.out.println("ArrayList size = " + colors.size());
    }
}

CodeRunner7_PopcornHack1.main(null);
```

## Popcorn Hack 2

```java
// CODE_RUNNER: Code

public class CodeRunner8_PopcornHack2 {

    String title;
    String author;
    int pages;

    public CodeRunner8_PopcornHack2(String title, String author, int pages) {
        this.title = title;
        this.author = author;
        this.pages = pages;
    }

    public String displayInfo() {
        return "Title: " + title +
               "\nAuthor: " + author +
               "\nPages: " + pages;
    }

    public boolean isLong() {
        return pages > 300;
    }

    public static void main(String[] args) {
        CodeRunner8_PopcornHack2 myBook =
            new CodeRunner8_PopcornHack2("Java Basics", "John Doe", 350);

        System.out.println(myBook.displayInfo());
        System.out.println("Is the book long? " + myBook.isLong());
    }
}

CodeRunner8_PopcornHack2.main(null);
```
