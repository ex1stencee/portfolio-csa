## Popcorn Hack

```java
public class Multiples {
    public static void main(String[] args) {
        int printed = 0;

        for (int i = 5; i <= 50; i += 5) {
            System.out.print(i + " ");
            printed++;
        }

        System.out.println();
        System.out.println("printed " + printed + " numbers");
    }
}

Multiples.main(null);
```

## Homework Hack

```java
public class Factorial {
    public static void main(String[] args) {
        int n = 5;
        int result = 1;

        for (int i = 1; i <= n; i++) {
            result = result * i;
        }

        System.out.println(n + "! = " + result);
    }
}

Factorial.main(null);
```
