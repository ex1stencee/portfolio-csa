## Popcorn Hack

We start by initializing `rows`, which will server as the number of rows. The first for loop counts from 1 up to the number of rows, establishing the rows that will be outputed. Then, the inner for loop makes sure that in each row the columns only count up to the value of the row, as shown by the for (int j = 1; j <= i; j++). The inner for loop then prints out the value and adds a space at the end for the next value. Once each inner loop is done, a `println()` is called to move onto the next line.

````java
public class NumberTriangle {
    public static void main(String[] args) {
        int rows = 4;
        int count = 0;

        for (int i = 1; i <= rows; i++) {

            for (int j = 1; j <= i; j++) {
                System.out.print(j + " ");
                count++;
            }

            System.out.println();
        }

        System.out.println("Iterations of inner body: " + count);
    }
}

NumberTriangle.main(null);
````

## Homework Hack

The code first initializes two variables: `n` and `pairs`. Pairs will serve as a counter for the number of pairs that the program fids. `n` will server as the maximum number that the program will check. We can see this with the line `for (int a = 1; a <= n; a++)` and `for (int b = a + 1; b <= n; b++)` where we see that the numbers will iterate up to the value `n` and then stop. Then, an if statement checks whether or not the sum of the values is even, and adds to pairs if it is not. Once the loops are done, the number of pairs is printed.

````java
public class EvenPairs {
    public static void main(String[] args) {
        int n = 4;
        int pairs = 0;
        int count = 0;

        for (int a = 1; a <= n; a++) {
            for (int b = a + 1; b <= n; b++) {
                count++;
                if ((a + b) % 2 == 0) {
                    pairs++;
                }
            }
        }

        System.out.println("Even-sum pairs: " + pairs);
        System.out.println("Iterations of inner body: " + count);
    }
}

EvenPairs.main(null);
````
