## Popcorn Hack 1

```java
public class Dog {
    public String Name;
    public int Age;

    public Dog(String name, int age) {
        Name = name;
        Age = age;
    }

    public String getName() {
        return Name;
    }

    public int getAge() {
        return Age;
    }

    public static void main(String[] args) {
        Dog d = new Dog("Larry", 4);

        System.out.println(d.getName());
        System.out.println(d.getAge());
    }
}

Dog.main(null);
```

## Popcorn Hack 2

```java
public class SubstringFinder {

    public static int findSubstring(String fullString, String target) {
        return fullString.indexOf(target);
    }

    public static void main(String[] args) {

        String sentence = "code code code security";

        int position = findSubstring(sentence, "security");

        System.out.println("Substring position: " + position);
    }
}

SubstringFinder.main(null);
```

## Popcorn Hack 3

```java
public class NumberGuessingGame {

    public static void main(String[] args) {

        // Generate a random number from 1 to 10
        int target = (int)(Math.random() * 10) + 1;

        // Example guess
        int guess = 7;

        // Calculate how close the guess is
        int difference = Math.abs(target - guess);

        System.out.println("guess: " + guess);
        System.out.println("target: " + target);
        System.out.println("off by: " + difference);

        if (guess == target) {
            System.out.println("you win!");
        } else {
            System.out.println("try again!");
        }
    }
}

NumberGuessingGame.main(null);
```
