## Popcorn Hack

 We start by establishing a variable `printed` to server as our counter. We then iterate using `for (int i = 5; i <= 50; i += 5)`. What this does is that it starts at 5, and then keeps adding 5 until it exceeds 50. This means that the loop will iterate for i = 5, 10, 15, 20, 25, 30, 35, 40, 45, 50. The inside of that loop then prints out the values of i.

````java
public class Multiples {
    public static void main(String[] args) {
        int printed = 0;

        for (int i = 5; i <= 50; i += 5) {
            System.out.print(i + " ");
            printed++;
        }

        System.out.println();
        System.out.println("printed " + printed + " numbers");
    }
}

Multiples.main(null);
````

While loop version. Same idea but uses while to check if the value is under 50 and i+=5 to count.

````java
public class Multiples {
    public static void main(String[] args) {
        int printed = 0;
        int i = 5;

        while (i <= 50) {
            System.out.print(i + " ");
            printed++;
            i += 5;
        }

        System.out.println();
        System.out.println("printed " + printed + " numbers");
    }
}

Multiples.main(null);
````

## Homework Hack

We start by establishing `n` as the number that will be factorialied, and establish a variable `result` to represent the result of the factorial. Since factorials work by multiplying every value lower than that number down to 1, we use a for loop like `for (int i = 1; i <= n; i++)` to iterate from 1 all the way up to the `n` and multiply result by that value. Thus, the value will be multiplied by every value between 1 and `n`, and we can print that value as the factorial.

````java
public class Factorial {
    public static void main(String[] args) {
        int n = 5;
        int result = 1;
        int printed = 0;

        for (int i = 1; i <= n; i++) {
            result = result * i;
            printed++;
        }

        System.out.println(n + "! = " + result);
        System.out.println("printed " + printed + " numbers");
    }
}

Factorial.main(null);
````

While loop version. Same idea, but using the while syntax to check if i <= n instead of a standard for loop.

````java
public class Factorial {
    public static void main(String[] args) {
        int n = 5;
        int result = 1;
        int printed = 0;
        int i = 1;

        // Repeat while i is less than or equal to n.
        while (i <= n) {
            result = result * i;
            printed++;
            i++;
        }

        System.out.println(n + "! = " + result);
        System.out.println("printed " + printed + " numbers");
    }
}

Factorial.main(null);
````
