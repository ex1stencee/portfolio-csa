## Popcorn Hack

First we define variables, one word that we will be testing, and a counter for the number of vowels. We then use a for loop to iterate for each character in the given word. For each value of i, we use the substring function to pull out the given letter, and use an if statement to determine whether or not the letter is a vowel. If the vowel is in fact a vowel, then the vowels counter would increase, with the finished output being the total number of vowels.

````java
public class VowelCounter {
    public static void main(String[] args) {
        String word = "iterate";
        int vowels = 0;

        // TODO: for each index i, take word.substring(i, i + 1)
        //       and count it if it equals a, e, i, o, or u
        for (int i = 0; i < word.length(); i++) {
            String letter = word.substring(i, i + 1);
            if (letter.equals("a") || letter.equals("e") ||
                letter.equals("i") || letter.equals("o") ||
                letter.equals("u")) {
                vowels++;
            }
        }

        System.out.println("Vowels: " + vowels);   // expected: 4
    }
}

VowelCounter.main(null);
````

## Homework Hack

We use a for loop to build the rev string, which counts from i to the length of the given string, and adds each character to the start of the string `rev`. Once the for loop finishes iterating all the characters of the string, `rev` becomes effectively the reversed string of the input, which we can compare with the original input to determine whether or not the string is truly a palindrome.

````java
public class Palindrome {
    public static boolean isPalindrome(String s) {
        String rev = "";
        // TODO: build rev by adding each character to the front
        for (int i = 0; i < s.length(); i++) {
            rev = s.substring(i, i+1) + rev;
        }

        // TODO: return whether rev equals s
        return rev.equals(s);
    }

    public static void main(String[] args) {
        System.out.println("racecar: " + isPalindrome("racecar"));   // true
        System.out.println("level: " + isPalindrome("level"));       // true
        System.out.println("java: " + isPalindrome("java"));         // false
    }
}

Palindrome.main(null);
````
