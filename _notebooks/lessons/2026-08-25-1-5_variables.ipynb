## Practice: Predict the Output

**Q1**
```java
int a = 10, b = 4; // Defines integer a as 10, and integer b as 4
System.out.println(a / b); // Calculates 10 / 4, trunactes, then prints [2]
System.out.println(a % b); // Calculates 10 mod 4, then prints [2]
System.out.println((double)(a / b)); // Calculates 10 / 4, trunactes, stores as a float, then prints [2.0]
System.out.println((double)a / b); // Stores a as a double, calculates 10.0 / 4, then prints [2.5]
```

## FRQ-Style Tasks

````java
public static double avgInt(int a, int b) {
    return (a + b) / 2.0;
}

public static double percent(int correct, int total) {
    return (double)correct / total;
}

public static int safeMod(int a, int b) {
    if (b == 0) {
        return 0;
    } else {
        return a % b;
    }
}

System.out.println(avgInt(9, 10));
System.out.println(percent(20, 23));
System.out.println(safeMod(24, 5));
````

## MCQ Practice Problems
**Question 1:** What is the output of the following code

```java
double x = 7.9;
int y = (int) x;
System.out.println(y);
```

A. 7.9\
B. 8\
C. 7\
D. 8.0\
E. Compile error

**Answer:** C. 7

**Explanation:** The first line stores double x as 7.9. The second line then converts the double x into an integer, and store as y, which would trunacte 7.9. Lastly, the program prints y, which is 7.

**Question 2:** Which of the following statements will compile without error?

```java
int a = 10;
double b = 5.5;
```

A. `int c = a + b;`\
B. `int c = (int)(a + b);`\
C. `int c = a + (int)b;`\
D. Both B and C\
E. All of the above

**Answer:** D. Both B and C

**Explanation:** Option A will error out since a + b is 15.5, which needs to be stored as a double, not an integer. Options B and C will both succeed since either b, the float, or the end result will be converted to an integer before being stored as an integer.

**Question 3:** What is the result of the following code?

```java
int num = 9 / 2;
System.out.println(num);
```

A. 4.5\
B. 4\
C. 5\
D. 4.0\
E. Compile error

**Answer:** B. 4

**Explanation:** Since we are doing integer division, the ending part will be trunacted, and thus the quotient 4.5 will instead be stored as 4 before being printed.

**Question 4:** Given the following code, what is printed?

A. 15
B. 16
C. 15.5
D. 16.0

**Answer:** B. 16

**Explanation:** First we store float 15.7 as double d. Then we add 0.5 to d, before converting it to an integer. This means that we will be trunacting the sum 16.2 to 16, before printing.

**Question 5:** Which statement about casting is TRUE?

A. Casting from int to double requires explicit casting\
B. Casting from double to int requires explicit casting\
C. Casting from int to double may result in loss of data\
D. Casting is never necessary in Java\
E. Casting from double to int always rounds to the nearest integer

**Answer:** B. Casting from double to int requires explicit casting

**Explanation:** Since we are losing data (going from a floating point to an integer), casting from a double to int requires explicit casting, while going the other way will happen automatically, since we are adding data. The final option, casting from double to int always rounds to the nearest integer is false since java trunactes.
